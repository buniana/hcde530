# A5 Homework — Exploring UX Tool Review Data

**Dataset:** `category_helpful_votes.csv` — 500 rows of UX tool reviews with a tool category and a helpful vote count.

**Questions answered:**
1. What does the dataset look like?
2. What's the distribution of helpful votes?
3. Which reviews were the most helpful (filter to a meaningful subset)?
4. Which UX tool category gets the most helpful votes on average?
5. Are there any missing values in the dataset?

```python
import pandas as pd

df = pd.read_csv('category_helpful_votes.csv')
```

## Question 1: What does the dataset look like?

```python
# What does this data actually contain? Checking the first few rows reveals the column names,
# value formats, and whether anything looks obviously wrong before running any analysis.
df.head()
```

```python
# How many rows and columns are there, and what type is each column?
# This confirms the dataset is the right size and that helpful_votes is numeric (needed for math later).
df.info()
```

**What this tells us:** The dataset has 500 rows and 2 columns — `category` (the type of UX tool being reviewed) and `helpful_votes` (how many people found the review helpful). Both columns are fully populated with no obvious issues.

## Question 2: What's the distribution of helpful votes?

```python
# How spread out are the helpful vote counts? Are most reviews low, medium, or high?
# A wide spread means the community is selective — some reviews get a lot of validation, others almost none.
df['helpful_votes'].value_counts().sort_index()
```

**What this tells us:** The helpful vote counts range from 0 to 50. Looking at the spread shows whether most reviews cluster around low, medium, or high helpfulness. A wide distribution suggests reviewers vary a lot in how useful others find their reviews.

## Question 3: Which reviews were the most helpful? (Filter to a meaningful subset)

```python
# Which reviews earned 40 or more helpful votes — the top tier of community validation?
# This isolates the reviews most people found genuinely useful, which reveals which tool categories
# tend to generate high-quality, trustworthy reviews.
highly_helpful = df[df['helpful_votes'] >= 40]
highly_helpful
```

**What this tells us:** This subset shows only the reviews that 40 or more people found helpful. These are the most community-validated reviews in the dataset. Looking at which categories appear here reveals which tool types tend to generate the most genuinely useful reviews.

## Question 4: Which UX tool category gets the most helpful votes on average?

```python
# Which tool category gets the highest average helpful votes?
# A higher average means that category's reviews tend to be more useful to readers —
# possibly because decisions about those tools carry higher stakes or attract more detailed reviewers.
df.groupby('category')['helpful_votes'].mean().sort_values(ascending=False)
```

**What this tells us:** This shows the average number of helpful votes per category, sorted from highest to lowest. Categories at the top tend to produce reviews that the community finds more valuable — which could reflect higher stakes in choosing those tools, more detailed review content, or a more engaged user base.

## Question 5: Are there any missing values in the dataset?

```python
# Is the dataset complete, or are there gaps that would affect the analysis?
# Zero missing values means we can use every row without dropping or filling anything.
df.isnull().sum()
```

**What this tells us:** A result of 0 for each column means the dataset is complete — there are no missing values in either `category` or `helpful_votes`. This means we can use the data as-is without needing to drop rows or fill in gaps before analysis.
